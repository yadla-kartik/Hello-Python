#### Deep Copy and Shallow Copy 

```python
# Shallow Copy - Makes a copy but inner parts are still connected to the original
```

```python
# Deep Copy - Makes a complete separate copy that won't affect the original
```

```python
ori = np.array([1,2,3,4,5,6])
ori
```

```python
slic = ori[:3]
slic[0] = 10
slic
```

```python
ori # Here we can se that when we slice from an ori array and do any changes in the sliced array then the changes will affect the original array. So, here the shallow copy is done as the inner parts are connected to the original
```

```python
deep = ori.copy()
deep
```

```python
deep[1] = 20
deep
```

```python
ori # Here we can see that the deep copy method is used. In deep copy a separate copy is created in the RAM both original and copied array are at different location that's why when changes are done at one array do not refelect other's
```

#### Matric Operations

```python
mt1 = np.arange(1,5).reshape(2,2)
mt2 = np.arange(5,9).reshape(2,2)
```

```python
mt1 + mt2
```

```python
mt1 * mt2  # This multiplies the matrix by index means, it doesn't uses matrix multiplication. To do multiplication using matric multiplication we use @ 
```

```python
 mt1 @ mt2
```

```python
mt1.dot(mt2) # The matrix multiplication is also known as a dot product so, there is function called dot 
```

```python
# Transpose Matrix
```

```python
mt1.T # Transpose the matrix mt1 (swap rows and columns)
```

#### Advance Array Manipulation 

##### Stacking Array

```python
# Stacking array - Means the joining and combining of multiple arrays to form a larger array. We cant stack the matrix array
```

```python
a = np.arange(6)
b = np.arange(6)
```

```python
np.vstack((a,b)) # Join Vertically 
```

```python
np.hstack((a,b)) # Join Horizontally
```

```python
np.column_stack((a,b))
```
