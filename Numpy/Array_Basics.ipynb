#### Random Generation Function

```python
np.random.rand(5) # This function only generates random numbers btw 0 to 1 so this is a concept of Normilization
```

```python
np.random.randn(5) # This function only generates random numbers btw -3 to 3 so this is a concept of Standardization
```

```python
np.random.randint(5,20,10) # This is a function where random integer number are generated randomly and in this array we will pass three arguments (Start No. , End No. , Number of Numbers you want to generate btw start and end)
```
