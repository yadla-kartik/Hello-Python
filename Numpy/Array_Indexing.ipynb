```python
import numpy as np
```

#### Numpy Indexing and Slicing of vectors

```python
arr = np.random.randint(0,10,10)
arr
```

```python
arr[1:4:1] # Array slicing is also same like a list, we are gonna pass for slicing the array are [start idx, end idx, steps]
```

#### Numpy Indexing and Slicing of Multi Dimensional 

```python
# In above we saw how the Indexing and Slicing works in vectors but now we will see how it will work in 2-D
```

```python
arr = np.random.randint(6,30,20).reshape(4,5)
arr
```

```python
arr[3] # this will give the whole row of the matrix. to access single element we gonna pass both row and column index
```

```python
arr[1,1] # see in this we passes 1,1 which means first 1 is the second row and second 1 is second column so the number is 17 
```

```python
arr[2,4]
```

```python
# Now we will se how slicing works in 2D Array
```

```python
arr = np.arange(0,15).reshape(3,5)
arr
```

```python
arr[0:2,1:4] # When we perform slicing for 2D array we have to pass both row and column index for slicing for ex - [Row-Start-Idx:End-Idx , Column-Start:End-Idx]
```

```python
arr[:2,:]
```

```python
arr[:,3] # This will give whole column 
```

#### Boolean Indexing

```python
arr = np.arange(0,10)
arr
```

```python
bool_idx = arr % 2  == 0
bool_idx
```

```python
arr = arr[bool_idx]
arr
```
