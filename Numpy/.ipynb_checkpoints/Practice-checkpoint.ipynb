```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
data.mean()
```

```python
#Extract Math marks of all students.
data[:,1]
```

```python
#Find the highest Science mark.
sci = data[:,2]
sci.max()
```

```python
#Get details of the student who scored more than 90 in Math.
math = data[:,1]
ans = math > 90 
result = data[ans]
result
```

```python
#Increase Math marks of all students by 5.
math + 5
```

```python
#Find how many students are younger than 19.
age = data[:,0]
filter_age = age < 19 
age = age[filter_age]
age
```

```python
#Calculate the average marks in each subject (column-wise mean).
math = data[:,1]
scii = data[:,2]
print(math.mean())
print(scii.mean())
```

```python
#Get data of students who scored at least 80 in both subjects.
anss = (math >= 80 ) & (scii >= 80)
resultsss = data[anss]
resultsss
```

```python
#Replace all Science marks < 75 with 0.
a = scii < 75
data[a,2] = 0
data
```

#### Valid Sudoku

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
a = np.sum(s,1)
for i in a:
    if(i != 45):
        print("Your suduko is not valid")
        break
else:
    print('The sum of row wise are equal')
```

```python
a = np.sum(s,0)
for i in a:
    if(i != 45):
        print("Your suduko is not valid")
        break
else:
    print('The sum of column wise are equal')
```

```python
for i in range(0,9,3):
    for j in range(0,9,3):
        n = s[i:i+3, j:j+3]
        print(n.sum())
```
