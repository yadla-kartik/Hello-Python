```python
import numpy as np
import pandas as pd
```

#### Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df.isna() # isna() function is used to find the missing values from a Dataframe
```

```python
df.isna().sum() # Sums the total no. of null values in each column
```

```python
df.isna().any() # Tells which column have null values. If a column have True then there is no null values
```

#### Removing Missing Data

```python
df.dropna() # Drops all the values row wise
```

```python
df.dropna(thresh=3) #  Remove rows that have fewer than 3 non-null values
```

#### Filling the Missing Data

```python
df
```

```python
df.fillna(0) # Replace all NaN/null values in the DataFrame with 0
```

```python
val = {'A':100,'B':200,'C':300,'D':400} 
df.fillna(value=val)  # Fill NaN values in each column with corresponding values from the dictionary. Filling is done by column wise
```
