## Pandas Series
 
A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.


#### Creating Series
There are multiple ways to create a pandas Series:

```python
import numpy as np
import pandas as pd

labels = [1,2,3]
mylist = ['a','b','c']
arr = np.array([10,20,30])
d = {'a':10, 'b':20, 'c':30}
```

```python
pd.Series(mylist, index=labels)
```

```python
pd.Series(d)
```

```python
pd.Series(arr, labels)
```

```python
pd.Series(mylist)
```
