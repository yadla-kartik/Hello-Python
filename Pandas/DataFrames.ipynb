```python
import numpy as np
import pandas as pd
```

#### Creating a DataFrame
A DataFrame can be created using two Data structures that are - 1. List and 2. Dictionary

```python
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "London", "Paris"]
}
# Dataframe is a two-dimensional, tabular data structure with labeled rows and columns. Also when we combine multiple Series then it becomes a Dataframe.
pd.DataFrame(data)
```

```python
data = [
    ["Alice", "Bob", "Charlie"],
    [25, 30, 35],
    ["New York", "London", "Paris"]
]
pd.DataFrame(data) # In this dataframe there is no header names like name age location, but we can that columns let's see in the below example
```

```python
col = ['Name', 'Age', 'City']
df = pd.DataFrame(data,columns=col) # By creating a separate header we can add into the Dataframe
df
```

#### Selection and Indexing of Columns

```python
# For targeting our columns or headers we use column names and you can use multiple names together
```

```python
df
```

```python
df['Name']
```

```python
df[['Name','City']]
```

#### Creating a new column

```python
df['Designation'] = ["Doctor", 'Engineer','MBBS'] 
df # We can create a new header 
```

#### Removing Columns

```python
df.drop('Designation', axis=1) #  Removes the 'Designation' column from the dataframe. axis=1 is used bcz 1 is for columns(Headers) and 0 is for rows. Drop function uses axis=0 by default.
```

```python
df # Here we can see that the dataframe still have the column designation it's bcz of we did'nt used the inplace params we have to give inplace=True then only the column will be deleted permanently from the dataframe.
```

```python
df.drop('Designation', axis=1, inplace=True) # This inplace true will delete the column permanently.
```

```python
df
```

#### Selection Rows

```python
data2 = {
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['IT', 'HR', 'IT', 'Sales', 'Sales'],
    'Age': [28, 34, 41, 29, 32],
    'Salary': [75000, 65000, 90000, 60000, 70000],
    'Years_Exp': [3, 7, 12, 4, 6],
    'Remote': [True, False, True, False, True]
}
df2 = pd.DataFrame(data2)
df2
```

```python
# To select our rows we will use the loc[3], It will give the whole row of index 3
```

```python
df2.loc[2]
```

```python
df2.loc[[1,2,3]] # We can also select multiple rows by providing multiple index
```

```python
df2.iloc[1]
```

#### Selection Subsets of Rows and Columns

```python
# To get a subset of a dataframe we use same rule like numpy. first we will pass the row index and then column header.
```

```python
df2.loc[[0,3]][['Age', 'Remote']]
```

#### Conditional Selection

```python
df2
```

```python
df2[df2['Age'] > 30] # Filter DataFrame to show only rows where Age column is greater than 30. Here we uses same rules like we used in Numpy
```

```python
df2[(df2['Age'] > 30 ) & (df2['Remote'] == True)] # Multiple conditions
```
