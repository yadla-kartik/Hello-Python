```python
import numpy as np
import pandas as pd
```

#### Group By

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.groupby('Category') # this code will return an object address. Returns a DataFrameGroupBy object (not the actual grouped data)
```

```python
# Group by Category and calculate the sum of Sales

cal = df.groupby('Category')['Sales'].sum() # Create a groupby object that groups the DataFrame by the 'Category' column
cal 
```

```python
# Group by multiple columns
# Group by Category and Store
cat = df.groupby(['Category','Store'])['Sales'].sum()
cat
```
