```python
import seaborn as sns
import matplotlib.pyplot as plt
```

#### Heat Map

```python
flights = sns.load_dataset('flights')
flights
```

```python
tips = sns.load_dataset("tips")
tips
```

```python
tipscorr = tips[['total_bill', 'size', 'tip']]
tipscorr
```

```python
tipscorr.corr() # We can find correlation using this function corr()
```

```python
sns.heatmap(tipscorr.corr(), annot= True)
```

#### Cluster Map

```python
sns.clustermap(tipscorr.corr())
```

#### Pivot table Heat Map

```python
flight_pvt = flights.pivot_table(values='passengers', index='month', columns='year')
flight_pvt
```

```python
sns.heatmap(flight_pvt)
```
